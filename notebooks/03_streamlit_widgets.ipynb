## 📚 Concept: How Widgets Work

Every Streamlit widget function (`st.button()`, `st.slider()`, etc.) has two jobs:

1. **Render** an interactive element in the browser.
2. **Return** a value to your Python script — the current state of that widget.

Because of the rerun model, every time *any* widget changes, the script reruns and every widget returns its latest value.

## 🧠 Intuition: Widgets Are Like Form Fields

Think of a Streamlit app as a **living form**:

```
Traditional Form          Streamlit App
┌──────────────┐         ┌──────────────────┐
│ Name: [    ] │         │ Name: [    ]     │
│ Age:  [    ] │   ═══>  │ Age:  [====]     │
│ Color:[ v ]  │         │ Color:[ v ]      │
│ [Submit]     │         │ (auto-updates!)  │
└──────────────┘         └──────────────────┘
   User clicks submit        Script reruns on
   to send data              every interaction
```

In a traditional form, you click "Submit" to send all values at once. In Streamlit, **every interaction is a mini-submit** — the script reruns immediately.

## 💡 Minimal Example: Every Widget Type

Save the following as `widgets_overview.py` and run with `streamlit run widgets_overview.py`:

````python
# === widgets_overview.py ===
# A complete overview of every core Streamlit widget.
# Save this file and run: streamlit run widgets_overview.py

widgets_code = '''
import streamlit as st
from datetime import date, time

st.set_page_config(page_title="Widgets Overview", page_icon="🎛️", layout="wide")
st.title("🎛️ Streamlit Widgets Overview")

# --- Buttons & Toggles ---
st.header("1. Buttons & Toggles")
col1, col2, col3 = st.columns(3)
with col1:
    if st.button("Click Me"):
        st.success("Button clicked!")
with col2:
    toggle = st.toggle("Dark mode")
    st.write(f"Dark mode: {toggle}")
with col3:
    agree = st.checkbox("I agree to terms")
    st.write(f"Agreed: {agree}")

st.divider()

# --- Selection Widgets ---
st.header("2. Selection Widgets")
col1, col2, col3, col4 = st.columns(4)
with col1:
    color = st.selectbox("Select a color", ["Red", "Green", "Blue"])
    st.write(f"Selected: {color}")
with col2:
    colors = st.multiselect("Pick colors", ["Red", "Green", "Blue", "Yellow"])
    st.write(f"Selected: {colors}")
with col3:
    option = st.radio("Choose", ["A", "B", "C"], horizontal=True)
    st.write(f"Radio: {option}")
with col4:
    size = st.select_slider("Size", options=["S", "M", "L", "XL"])
    st.write(f"Size: {size}")

st.divider()

# --- Text Input ---
st.header("3. Text Input")
col1, col2 = st.columns(2)
with col1:
    name = st.text_input("Name", placeholder="Enter your name")
with col2:
    notes = st.text_area("Notes", placeholder="Enter notes here...", height=100)

st.divider()

# --- Number & Slider ---
st.header("4. Number & Slider")
col1, col2, col3 = st.columns(3)
with col1:
    age = st.number_input("Age", 0, 120, 25)
with col2:
    score = st.slider("Score", 0, 100, 75)
with col3:
    range_val = st.slider("Range", 0, 100, (20, 80))
    st.write(f"Range: {range_val}")

st.divider()

# --- Date & Time ---
st.header("5. Date & Time")
col1, col2 = st.columns(2)
with col1:
    dob = st.date_input("Date of birth", value=date(2000, 1, 1))
with col2:
    meeting = st.time_input("Meeting time", value=time(9, 0))

st.divider()

# --- Color Picker ---
st.header("6. Color Picker")
color = st.color_picker("Pick a brand color", "#FF5733")
st.markdown(f"<div style='background:{color};height:50px;border-radius:8px;'></div>", unsafe_allow_html=True)

st.divider()
st.caption("Widgets Overview • Module 02 • Streamlit for Data Science")
'''

print("Save the code above as widgets_overview.py and run:\n")
print("  streamlit run widgets_overview.py\n")
print("Expected output:")
print("  - A page with 6 sections, each demonstrating different widget types")
print("  - All widgets are interactive — values update as you interact")
print("  - Layout uses st.columns() for side-by-side display")
````

## 🏗️ Data Science Example: Dataset Explorer

Here's a practical Data Science app that uses widgets to explore a dataset interactively.

````python
# === dataset_explorer.py ===
# A practical Data Science app with interactive widget controls.
# Save this file and run: streamlit run dataset_explorer.py

explorer_code = '''
import streamlit as st
import pandas as pd
import numpy as np

st.set_page_config(page_title="Dataset Explorer", page_icon="📊", layout="wide")
st.title("📊 Dataset Explorer")
st.markdown("Interactively explore a synthetic student dataset using widget controls.")

st.divider()

# --- Generate Data ---
np.random.seed(42)
n = 200
df = pd.DataFrame({
    "student_id": range(1, n + 1),
    "name": [f"Student {i}" for i in range(1, n + 1)],
    "age": np.random.randint(18, 25, n),
    "gpa": np.random.uniform(2.0, 4.0, n).round(2),
    "department": np.random.choice(["CS", "Math", "Physics", "Biology"], n),
    "credits": np.random.randint(60, 180, n),
    "scholarship": np.random.choice([True, False], n, p=[0.3, 0.7]),
})

# --- Sidebar Filters ---
st.sidebar.header("🔍 Filters")

departments = st.sidebar.multiselect(
    "Departments", df["department"].unique(), default=df["department"].unique()
)

min_gpa, max_gpa = st.sidebar.slider("GPA Range", 0.0, 4.0, (2.0, 4.0))

min_credits = st.sidebar.number_input("Minimum credits", 0, 200, 60)

scholarship_only = st.sidebar.checkbox("Scholarship holders only")

# --- Apply Filters ---
filtered = df[
    (df["department"].isin(departments)) &
    (df["gpa"] >= min_gpa) &
    (df["gpa"] <= max_gpa) &
    (df["credits"] >= min_credits)
]

if scholarship_only:
    filtered = filtered[filtered["scholarship"] == True]

# --- Metrics ---
st.header("📈 Summary")
col1, col2, col3, col4 = st.columns(4)
col1.metric("Students", len(filtered))
col2.metric("Avg GPA", f"{filtered['gpa'].mean():.2f}" if len(filtered) > 0 else "N/A")
col3.metric("Avg Credits", f"{filtered['credits'].mean():.0f}" if len(filtered) > 0 else "N/A")
col4.metric("Scholarship %", f"{filtered['scholarship'].mean() * 100:.1f}%" if len(filtered) > 0 else "N/A")

st.divider()

# --- Data Table ---
st.header("📋 Filtered Data")
st.dataframe(filtered, use_container_width=True)

st.divider()
st.caption(f"Showing {len(filtered)} of {len(df)} students • Dataset Explorer • Module 02")
'''

print("Save the code above as dataset_explorer.py and run:\n")
print("  streamlit run dataset_explorer.py\n")
print("Features demonstrated:")
print("  - st.sidebar.multiselect() for department filtering")
print("  - st.sidebar.slider() for GPA range")
print("  - st.sidebar.number_input() for credit threshold")
print("  - st.sidebar.checkbox() for boolean filter")
print("  - st.metric() for KPI cards")
print("  - st.dataframe() for interactive table")
````

## 🔨 Build It: Complete App with Forms

Let's build a form-based input app. Save as `student_form.py`:

````python
# === student_form.py ===
# Demonstrates st.form() for batch widget submission.
# Save this file and run: streamlit run student_form.py

form_code = '''
import streamlit as st
from datetime import date

st.set_page_config(page_title="Student Form", page_icon="📝")
st.title("📝 Student Registration Form")
st.markdown("Fill out the form below. Values are submitted only when you click Submit.")

st.divider()

# --- Form ---
with st.form("student_form"):
    col1, col2 = st.columns(2)
    
    with col1:
        first_name = st.text_input("First Name *", placeholder="e.g., Ali")
        last_name = st.text_input("Last Name *", placeholder="e.g., Khan")
        email = st.text_input("Email *", type="email")
    
    with col2:
        dob = st.date_input("Date of Birth")
        department = st.selectbox("Department", ["CS", "Math", "Physics", "Biology"])
        gpa = st.slider("GPA", 0.0, 4.0, 3.0, step=0.1)
    
    notes = st.text_area("Additional Notes", placeholder="Optional...")
    
    submitted = st.form_submit_button("Submit Registration")

# --- Handle Submission ---
if submitted:
    if not first_name or not last_name or not email:
        st.error("Please fill in all required fields (First Name, Last Name, Email).")
    else:
        st.success("Registration submitted successfully! ✅")
        st.divider()
        st.header("📋 Submitted Data")
        col1, col2 = st.columns(2)
        with col1:
            st.write(f"**Name:** {first_name} {last_name}")
            st.write(f"**Email:** {email}")
            st.write(f"**DOB:** {dob}")
        with col2:
            st.write(f"**Department:** {department}")
            st.write(f"**GPA:** {gpa}")
        if notes:
            st.write(f"**Notes:** {notes}")
'''

print("Save the code above as student_form.py and run:\n")
print("  streamlit run student_form.py\n")
print("Key concepts:")
print("  - st.form() groups widgets for batch submission")
print("  - st.form_submit_button() triggers form submission")
print("  - Widgets inside forms don't trigger reruns on interaction")
print("  - Basic validation: check required fields before processing")
````

## ⚠️ Common Mistakes

| # | Mistake | Error/Symptom | Fix |
|---|---|---|---|
| 1 | Missing `key` parameter | Widget value lost on reorder | Add explicit `key` to every widget |
| 2 | Widget inside `if` block | Widget not always rendered | Use `disabled` parameter instead |
| 3 | Setting session_state after widget | Widget doesn't update | Set value *before* the widget renders |
| 4 | Nested forms | `StreamlitAPIException` | Forms can't be nested |
| 5 | Expecting rerun in form | Widget changes don't update output | Only `form_submit_button` triggers rerun |
| 6 | Using `validate` on old Streamlit | Parameter not recognized | Upgrade to Streamlit ≥ 1.62 |
| 7 | Confusing return value vs session_state | Cross-widget logic fails | Use session_state for linked widgets |

---

## 🐛 Debugging Tips

| Symptom | Likely Cause | Solution |
|---|---|---|
| Widget value resets | No `key` parameter | Add `key="my_widget"` |
| Widget not updating | Setting session_state after widget | Move assignment above widget |
| Form submits on every change | Widget outside form block | Move widget inside `with st.form()` |
| `validate` not working | Streamlit < 1.62 | Check version with `streamlit version` |
| `StreamlitAPIException` | Nested forms or duplicate keys | Use unique keys, don't nest forms |
| Widget appears disabled | `disabled=True` | Remove disabled parameter |
| Sidebar not showing | Missing sidebar calls | Use `st.sidebar.widget()` |

---

## ✅ Best Practices

1. **Always use explicit `key` parameters** — avoid auto-generated keys.
2. **Use descriptive key names** — `key="email"` not `key="ti1"`.
3. **Keep keys unique** — no two widgets share a key.
4. **Use sidebar for filters** — keeps main area clean.
5. **Use forms for batch input** — avoids per-keystroke reruns.
6. **Validate before processing** — check required fields first.
7. **Use `st.columns()` for layout** — side-by-side widgets are cleaner.
8. **Add `st.divider()` between sections** — visual separation helps.

## 📌 Key Takeaways

1. **Every widget returns a value** on every rerun — that's how Streamlit gets input.
2. **Selection widgets**: `selectbox` (single), `multiselect` (multiple), `radio` (exclusive).
3. **Text input**: `text_input` (single-line), `text_area` (multi-line), supports `validate` (≥ 1.62).
4. **Numeric input**: `number_input` (precise), `slider` (visual range).
5. **Date/time**: `date_input`, `time_input` (revamped in 1.61).
6. **Forms** group widgets for batch submission — no per-keystroke reruns.
7. **Always use explicit `key` parameters** for stability and session_state access.
8. **Choose the right widget** for the task — see the widget selection table in the reading.
