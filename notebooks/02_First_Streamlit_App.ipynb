## 🧠 Intuition: The Rerun Model in Action

Imagine you have a script that:
1. Generates a random number
2. Displays it

```python
import streamlit as st
import random

st.write(f"Random number: {random.randint(1, 100)}")
```

Every time you interact with *any* widget (even one that doesn't affect this code), the script reruns and a **new random number** appears. This is the rerun model in action.

### Why This Matters

| Without Rerun Model | With Rerun Model |
|---|---|
| You define routes manually | The whole script is the "route" |
| State is managed by the server | State is in `st.session_state` |
| You control when output updates | Output updates on every interaction |
| More boilerplate | Less boilerplate, but requires new thinking |

## 💡 Example: Anatomy of a Streamlit App

Here's the structure every Streamlit app follows:

````python
# === ANATOMY OF A STREAMLIT APP ===
# This code won't run in a notebook — it's meant to be saved as a .py file.
# We'll walk through each section below.

app_structure = """
# 1. IMPORTS
import streamlit as st
import pandas as pd

# 2. PAGE CONFIG (must be first st. call!)
st.set_page_config(
    page_title="My First App",
    page_icon="🚀",
    layout="centered",
)

# 3. TITLE AND DESCRIPTION
st.title("🚀 My First Streamlit App")
st.markdown("Welcome to my app! This is a simple demo.")

# 4. DATA
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Score": [92, 87, 95],
})

# 5. DISPLAY
st.dataframe(df)

# 6. INTERACTIVITY (we'll learn widgets in Module 02)
if st.button("Show Statistics"):
    st.write(df.describe())
"""

print(app_structure)
````

### The Six Sections

| # | Section | Purpose |
|---|---|---|
| 1 | **Imports** | Load Streamlit and other libraries |
| 2 | **Page Config** | Set browser tab title, icon, layout — **must be first** |
| 3 | **Title & Description** | Tell the user what the app does |
| 4 | **Data** | Load, generate, or compute data |
| 5 | **Display** | Render data with `st.dataframe()`, `st.write()`, charts |
| 6 | **Interactivity** | Widgets for user input (covered in Module 02) |

## 🔨 Build It: Text Elements

Let's learn each text-rendering function. Save the following as `text_demo.py` and run with `streamlit run text_demo.py`.

````python
# === text_demo.py ===
# Save this file and run: streamlit run text_demo.py

demo_code = '''
import streamlit as st

st.set_page_config(page_title="Text Demo", page_icon="📝", layout="centered")

# --- Titles ---
st.title("📝 Text Elements Demo")
st.caption("A tour of Streamlit's text-rendering capabilities")

# --- Headers ---
st.header("1. Headers")
st.write("Streamlit provides three levels of section headers:")
st.subheader("This is a subheader (h3)")
st.markdown("**Bold text** and *italic text* are supported via Markdown.")

st.divider()

# --- Markdown ---
st.header("2. Markdown")
st.markdown("""
- **Bold** text
- *Italic* text
- `Inline code`
- [Link to Streamlit](https://streamlit.io)
- ~~Strikethrough~~
""")

st.divider()

# --- Code Blocks ---
st.header("3. Code Blocks")
st.code("""import pandas as pd
df = pd.read_csv('sales.csv')
st.dataframe(df)""", language="python")

st.divider()

# --- Captions ---
st.header("4. Captions")
st.write("This is normal text.")
st.caption("This is a small gray caption — great for footnotes and annotations.")

st.divider()

# --- Dividers ---
st.header("5. Dividers")
st.write("Content above the divider.")
st.divider()
st.write("Content below the divider.")

st.divider()

# --- LaTeX ---
st.header("6. LaTeX")
st.latex(r"E = mc^2")
st.latex(r"\\int_{-\\infty}^{\\infty} e^{-x^2} dx = \\sqrt{\\pi}")
'''

print("Save the code above as text_demo.py and run:\n")
print("  streamlit run text_demo.py\n")
print("Expected output: A styled web page with headers, markdown, code blocks, captions, dividers, and LaTeX.")
````

## 🔬 Experiment: The Rerun Trap

This is the most important experiment in the course. It demonstrates why the rerun model matters.

````python
# === Experiment: Rerun Trap ===
# Save as rerun_trap.py and run with streamlit

rerun_trap = '''
import streamlit as st
import random

st.title("🎲 Rerun Trap Demo")
st.markdown("""**The Problem:** This counter always shows 0 or 1.

Every time you click "Increment", the entire script reruns.
`count = 0` resets on every rerun.""")

# BUG: count resets on every rerun!
count = 0
if st.button("Increment (Buggy)"):
    count += 1
st.write(f"Count: {count}")

st.divider()

# FIX: Use session_state to persist across reruns
st.markdown("""**The Fix:** Use `st.session_state` to persist values.""")

if "count" not in st.session_state:
    st.session_state.count = 0

if st.button("Increment (Fixed)"):
    st.session_state.count += 1
st.write(f"Count: {st.session_state.count}")
'''

print("Save the code above as rerun_trap.py and run:\n")
print("  streamlit run rerun_trap.py\n")
print("Observe:")
print("  - The buggy counter resets to 0 or 1 every click.")
print("  - The fixed counter increments correctly: 1, 2, 3, ...")
print("  - This is because st.session_state persists across reruns.")
````

## 🔨 Build It: Complete First App

Let's build a complete, working Streamlit app. Save this as `first_app.py`:

````python
# === first_app.py ===

first_app = '''
import streamlit as st
import pandas as pd
import numpy as np

# 1. PAGE CONFIG
st.set_page_config(
    page_title="Student Grade Analyzer",
    page_icon="🎓",
    layout="centered",
)

# 2. TITLE AND DESCRIPTION
st.title("🎓 Student Grade Analyzer")
st.markdown("""
Welcome to the **Student Grade Analyzer**!

This app demonstrates basic Streamlit text and data display.
We'll generate sample student data and show it interactively.
""")

st.divider()

# 3. GENERATE SAMPLE DATA
np.random.seed(42)
n_students = 30

data = pd.DataFrame({
    "Student": [f"Student {i+1}" for i in range(n_students)],
    "Math": np.random.randint(50, 100, n_students),
    "Science": np.random.randint(50, 100, n_students),
    "English": np.random.randint(50, 100, n_students),
})
data["Average"] = data[["Math", "Science", "English"]].mean(axis=1).round(1)
data["Grade"] = data["Average"].apply(
    lambda x: "A" if x >= 90 else "B" if x >= 80 else "C" if x >= 70 else "D" if x >= 60 else "F"
)

# 4. DISPLAY DATA
st.header("📊 Student Grades")
st.dataframe(data, use_container_width=True)

st.divider()

# 5. SUMMARY STATISTICS
st.header("📈 Summary Statistics")
st.write(data[["Math", "Science", "English", "Average"]].describe().round(1))

st.divider()

# 6. GRADE DISTRIBUTION
st.header("🎯 Grade Distribution")
grade_counts = data["Grade"].value_counts().sort_index()
st.bar_chart(grade_counts)

st.divider()
st.caption("Built with Streamlit • Module 01 • Student Grade Analyzer")
'''

print("Save the code above as first_app.py and run:\n")
print("  streamlit run first_app.py\n")
print("Expected output:")
print("  - A styled page with title and description")
print("  - An interactive DataFrame of 30 students")
print("  - Summary statistics (mean, std, min, max)")
print("  - A bar chart of grade distribution")
````

## 🧪 Experiment: `st.write()` Polymorphism

`st.write()` is Streamlit's Swiss Army knife. It accepts many data types and renders them appropriately.

````python
# Demonstrate st.write() polymorphism
# In a Streamlit app, each st.write() call renders differently based on input type.

print("=== st.write() Polymorphism ===")
print()

# String → paragraph
print('st.write("Hello, world!")  →  Renders as <p>Hello, world!</p>')
print()

# DataFrame → interactive table
df = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
print("st.write(df)             →  Renders as interactive DataFrame table")
print()

# Dict → JSON
print('st.write({"key": "val"})  →  Renders as JSON')
print()

# Matplotlib figure → image
print("st.write(fig)            →  Renders Matplotlib figure as image")
print()

# Plotly figure → interactive chart
print("st.write(fig)            →  Renders Plotly figure as interactive chart")
print()

# Multiple arguments
print('st.write("Hello", df)     →  Renders both the string and the DataFrame')
print()
print("Tip: For explicit control, use st.dataframe(), st.write(), st.markdown(), etc.")
````

## 🏗️ Practical Example: Quick Data Profile App

Here's a complete, Data-Science-oriented Streamlit app that profiles a dataset:

````python
# === data_profile.py ===
# A practical Data Science Streamlit app

data_profile_app = '''
import streamlit as st
import pandas as pd
import numpy as np

st.set_page_config(page_title="Data Profile", page_icon="🔍", layout="wide")

st.title("🔍 Quick Data Profile")
st.markdown("Upload a CSV or use sample data to get an instant profile.")

st.divider()

# --- Sample Data ---
np.random.seed(42)
sample_data = pd.DataFrame({
    "age": np.random.randint(18, 65, 100),
    "income": np.random.normal(50000, 15000, 100).round(0),
    "score": np.random.uniform(0, 100, 100).round(1),
    "category": np.random.choice(["A", "B", "C"], 100),
})

st.header("📊 Dataset Overview")
col1, col2, col3, col4 = st.columns(4)
col1.metric("Rows", len(sample_data))
col2.metric("Columns", len(sample_data.columns))
col3.metric("Numeric Cols", len(sample_data.select_dtypes(include=np.number).columns))
col4.metric("Categorical Cols", len(sample_data.select_dtypes(include="object").columns))

st.divider()

st.header("📋 Data Preview")
st.dataframe(sample_data.head(20), use_container_width=True)

st.divider()

st.header("📈 Descriptive Statistics")
st.dataframe(sample_data.describe().round(2), use_container_width=True)

st.divider()

st.header("📊 Distributions")
numeric_cols = sample_data.select_dtypes(include=np.number).columns.tolist()
selected_col = st.selectbox("Select a column to visualize:", numeric_cols)
st.bar_chart(sample_data[selected_col].value_counts().sort_index())

st.divider()
st.caption("Quick Data Profile • Built with Streamlit")
'''

print("Save the code above as data_profile.py and run:\n")
print("  streamlit run data_profile.py\n")
print("Features:")
print("  - st.columns() for metric cards")
print("  - st.metric() for KPI display")
print("  - st.dataframe() for interactive tables")
print("  - st.selectbox() for column selection")
print("  - st.bar_chart() for distributions")
````

## ⚠️ Common Mistakes

| # | Mistake | Error/Symptom | Fix |
|---|---|---|---|
| 1 | `st.set_page_config()` not first | `StreamlitAPIException` | Move it right after imports |
| 2 | Using `print()` for output | Output in terminal, not browser | Use `st.write()` |
| 3 | Variable resets on rerun | Counter always shows 0 | Use `st.session_state` |
| 4 | `st.set_page_config()` after conditional | Error when condition is true | Make it unconditional |
| 5 | Importing inside functions | Slow startup, repeated imports | Import at the top |
| 6 | No virtual environment | Dependency conflicts | Always use venv |

---

## 🐛 Debugging Tips

| Symptom | Likely Cause | Solution |
|---|---|---|
| `ModuleNotFoundError: No module named 'streamlit'` | Wrong env | Activate the correct venv |
| `StreamlitAPIException` about page config | Not first `st.` call | Move `st.set_page_config()` to the top |
| App doesn't update after code change | Browser cached | Hard refresh (Ctrl+Shift+R) |
| Widget value disappears | Not using session_state | Store in `st.session_state` |
| `Port 8501 is already in use` | Another Streamlit app running | Use `--server.port 8502` |
| App runs but nothing shows | Missing `st.` calls | Ensure you have `st.write()` or similar |

---

## ✅ Best Practices

1. **`st.set_page_config()` first, always.** No exceptions.
2. **Use `st.session_state` for persistence.** Don't rely on variable assignment.
3. **Keep imports at the top.** Don't import inside functions or loops.
4. **Use `st.write()` for quick debugging.** It works with almost any data type.
5. **Use specific functions when possible.** `st.dataframe()` over `st.write(df)` for clarity.
6. **Add `st.divider()` between sections.** Visual separation improves readability.
7. **Use `st.caption()` for footnotes.** It's cleaner than small text.
8. **Version control your apps.** They're just `.py` files — Git works perfectly.

## 📌 Key Takeaways

1. **Install** Streamlit with `pip install streamlit` in a virtual environment.
2. **Run** an app with `streamlit run app.py` — it opens at `localhost:8501`.
3. **Rerun model**: the entire script reruns on every interaction — this is fundamental.
4. **`st.session_state`**: the fix for persisting data across reruns.
5. **`st.set_page_config()`**: must be the first `st.` call, unconditionally.
6. **Text elements**: `st.title`, `st.header`, `st.subheader`, `st.write`, `st.markdown`, `st.caption`, `st.code`, `st.divider`, `st.latex`.
7. **`st.write()` is polymorphic**: strings, DataFrames, dicts, charts — it renders them all.
8. **App structure**: imports → page config → title → data → display → interactivity.
