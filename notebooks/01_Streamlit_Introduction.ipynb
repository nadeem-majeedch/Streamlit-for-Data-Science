## 📚 Concept: What Is Streamlit?

Streamlit is an **open-source Python framework** for building interactive data applications. 
It lets you create web apps by writing plain Python — no HTML, CSS, or JavaScript required.

### The Problem Streamlit Solves

Data scientists face a persistent gap:

```
Notebook (exploration)  →  ???  →  Deployed App (sharing)
         Private              Unknown              Public
```

Traditionally, bridging this gap requires:
- A front-end framework (React, Vue)
- A back-end API (Flask, FastAPI)
- DevOps knowledge (Docker, cloud deployment)

Streamlit eliminates all of that. You write a Python script, and Streamlit renders it as an interactive web application.

## 🧠 Intuition: The Notebook-to-App Pipeline

Think of Streamlit as a **"supercharged Jupyter cell"** that:

1. **Reruns automatically** whenever the user interacts with it.
2. **Renders to a browser** instead of a notebook output cell.
3. **Can be deployed** with one click to the cloud.

```
  Jupyter Notebook                    Streamlit App
  ┌────────────────┐                ┌──────────────────────┐
  │ df = pd.read.. │                │ df = pd.read..       │
  │ st.write(df)   │   ─────────►  │ st.dataframe(df)     │
  │ fig = px...    │                │ st.plotly_chart(fig)  │
  │ fig.show()     │                │                      │
  │ (private)      │                │ (shared via URL)     │
  └────────────────┘                └──────────────────────┘
```

The **code is almost identical** — you mostly replace `fig.show()` with `st.plotly_chart(fig)`.

## 🖼 Visual Explanation: The Execution Model

Streamlit's most important concept is the **rerun model**:

```
User clicks button
        │
        ▼
┌───────────────────────┐
│ Streamlit reruns the  │
│ ENTIRE script from    │
│ top to bottom         │
│                       │
│ 1. Import libraries   │
│ 2. st.set_page_config │
│ 3. Load / compute     │
│ 4. Render widgets     │
│ 5. Display output     │
└───────────────────────┘
        │
        ▼
   Browser updates
```

This is fundamentally different from a traditional web app where you define routes and handlers. In Streamlit, **the entire script is the handler**.

## ⚠️ Common Mistakes

| Mistake | Why It Happens | Fix |
|---|---|---|
| Confusing Streamlit with Flask | Different paradigms | Streamlit reruns the whole script; Flask uses routes |
| Thinking Streamlit replaces Jupyter | They serve different purposes | Use Jupyter for exploration, Streamlit for sharing |
| Assuming Streamlit is just for dashboards | It's more flexible | Streamlit builds full interactive apps, not just dashboards |
| Ignoring the rerun model |习惯了Jupyter的cell-by-cell execution | Always think "what happens when the script reruns?" |

---

## 🐛 Debugging Tips

| Symptom | Likely Cause | Solution |
|---|---|---|
| "Streamlit not found" | Not installed in active env | Run `pip install streamlit` |
| App doesn't open | Browser blocked popup | Manually go to `http://localhost:8501` |
| Import errors | Wrong virtual environment | Activate the correct venv |
| Slow rendering | No caching on expensive ops | Add `@st.cache_data` (Module 07) |

---

## ✅ Best Practices

1. **Always use a virtual environment** — prevents dependency conflicts.
2. **Start simple** — get `st.write("hello")` working before adding complexity.
3. **Think in reruns** — every line executes on every interaction.
4. **Use `st.session_state`** for anything that must persist (covered in M05).
5. **Version control** — Streamlit apps are just `.py` files; Git works perfectly.
6. **Separate data logic** — put heavy computation in functions, not at module level.

## 📌 Key Takeaways

1. **Streamlit = Python scripts → interactive web apps** (no front-end code needed).
2. **Rerun model**: the entire script reruns on every interaction — this is the #1 concept to internalize.
3. **Best for**: dashboards, ML demos, data exploration apps, prototyping.
4. **Not ideal for**: custom UIs, high-traffic APIs, pixel-perfect design.
5. **Compare to**: Jupyter (exploration), Flask (APIs), Dash (complex dashboards), Gradio (ML demos).
